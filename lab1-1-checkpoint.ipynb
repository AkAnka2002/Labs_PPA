```python
model_x_answers = {}
for question in lab1x1.answers:
    for model in lab1x1.answers[question]:
        if model in model_x_answers:
            model_x_answers[model] = model_x_answers[model] + [lab1x1.answers[question][model]]
        else:
            model_x_answers[model] = [lab1x1.answers[question][model]]
```

```python
import bert_score
```

```python
metrics = []
i = 1
for model_cand in model_x_answers:
    for model_ref in model_x_answers:
        print(f"[{i}/{4*4}] {model_cand} x {model_ref}")
        P, R, F1 = bert_score.score(model_x_answers[model_cand], model_x_answers[model_ref], lang="ru", verbose=False)
        res = {"model_cand": model_cand, "model_ref": model_ref, "precision": P, "recall": R, "F1": F1}
        metrics.append(res)
        print(res)
        i+=1
```

```python
import numpy as np
import matplotlib.pyplot as plt
from math import pi

# Data
categories = ['актуальность', 'корректность', 'креативность', 
              'воспроизводимость', 'скорость', 'стоимость', 'безопасность']
values = {
    'GPT': [10, 9, 9, 9, 8, 7, 9],
    'Grok': [8, 9, 7, 8, 7, 8, 9],
    'Perplexity': [9, 7, 8, 6, 6, 7, 7],
    'GigaChat': [6, 7, 5, 6, 9, 9, 8]
}

N = len(categories)
angles = [n / float(N) * 2 * pi for n in range(N)]
angles += angles[:1]

# Индивидуальные цвета для каждой модели (можно менять)
model_colors = {
    'GPT': '#FF6B6B',      # Коралловый красный
    'Grok': '#4ECDC4',     # Бирюзовый
    'Perplexity': '#45B7D1', # Голубой
    'GigaChat': '#96CEB4'   # Мятный
}

# Альтернативные цветовые схемы (раскомментируйте нужную)
# model_colors = {
#     'GPT': '#E63946',      # Насыщенный красный
#     'Grok': '#457B9D',     # Темно-синий
#     'Perplexity': '#2A9D8F', # Зеленый
#     'GigaChat': '#E9C46A'    # Желтый
# }

# Вариант 2: Пастельные тона
# model_colors = {
#     'GPT': '#FFB3BA',      # Светло-розовый
#     'Grok': '#B5EAD7',     # Мятно-зеленый
#     'Perplexity': '#C7CEEA', # Лавандовый
#     'GigaChat': '#FFDAC1'    # Персиковый
# }

# Создаем отдельную диаграмму для каждой модели
for model_name, values_list in values.items():
    fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True))
    
    # Подготавливаем данные (замыкаем круг)
    vals = values_list + values_list[:1]
    
    # Цвет для текущей модели
    color = model_colors[model_name]
    
    # Рисуем радар
    ax.plot(angles, vals, linewidth=2.5, linestyle='solid', color=color, marker='o', markersize=6)
    ax.fill(angles, vals, alpha=0.25, color=color)
    
    # Настройка осей
    ax.set_xticks(angles[:-1])
    ax.set_xticklabels(categories, fontsize=11, fontweight='bold')
    ax.set_rlabel_position(30)
    plt.yticks([2, 4, 6, 8, 10], ["2", "4", "6", "8", "10"], color="grey", size=9)
    ax.set_ylim(0, 10)
    
    # Добавляем сетку
    ax.grid(True, alpha=0.3)
    
    # Заголовок
    plt.title(f'Модель: {model_name}', size=16, pad=20, fontweight='bold')
    
    # Добавляем значения на график
    for i, (angle, val) in enumerate(zip(angles[:-1], values_list)):
        ax.text(angle, val + 0.5, str(val), 
                ha='center', va='center', 
                fontsize=10, fontweight='bold', 
                color=color)
    
    # Сохраняем или показываем
    plt.tight_layout()
    # plt.savefig(f'radar_{model_name}.png', dpi=300, bbox_inches='tight')  # сохранить каждый
    plt.show()

# Дополнительно: создаем субплоты (2x2 сетка)
fig, axes = plt.subplots(2, 2, figsize=(12, 12), subplot_kw=dict(polar=True))
axes = axes.flatten()

for idx, (model_name, values_list) in enumerate(values.items()):
    ax = axes[idx]
    
    vals = values_list + values_list[:1]
    color = model_colors[model_name]
    
    ax.plot(angles, vals, linewidth=2, color=color, marker='o', markersize=5)
    ax.fill(angles, vals, alpha=0.2, color=color)
    
    ax.set_xticks(angles[:-1])
    ax.set_xticklabels(categories, fontsize=9)
    ax.set_ylim(0, 10)
    ax.set_yticks([2, 4, 6, 8])
    ax.set_yticklabels(['2', '4', '6', '8'], fontsize=8)
    ax.grid(True, alpha=0.3)
    ax.set_title(model_name, size=12, fontweight='bold', pad=10)
    
    # Добавляем значения
    for i, (angle, val) in enumerate(zip(angles[:-1], values_list)):
        ax.text(angle, val + 0.5, str(val), 
                ha='center', va='center', fontsize=8, color=color)

plt.suptitle('Сравнение AI моделей — по отдельности', size=16, y=1.02)
plt.tight_layout()
plt.show()
# plt.savefig('radar_all_models_subplots.png', dpi=300, bbox_inches='tight')
```
